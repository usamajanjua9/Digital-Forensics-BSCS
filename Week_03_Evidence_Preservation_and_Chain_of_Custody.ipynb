## 🔹 1. Digital Evidence Preservation

---

### **1.1 What Is Preservation?**

**Definition:**

Preservation means protecting relevant digital material from avoidable change, loss, or contamination so later examination remains meaningful and reviewable.

💡 **In simple words:** *Freeze the story in the records. Do not let careless handling rewrite the evidence before anyone examines it.*

**Why this idea sits at the front of the week:**
Almost every later forensic technique assumes a trustworthy baseline. Imaging, hashing campaigns, timelines, and reports all inherit whatever happened during the first hours of control. Preservation is therefore not a side note - it is the condition that makes later technical work interpretable.

**An investigator asks:**

* What item is relevant (USB, export, log package, laptop)?
* What change risks exist if someone casually opens it?
* Who is authorized to receive and store it?
* What becomes the preserved baseline (device, image, or export set)?
* What will analysts open instead (working copy)?
* What must be written down now so a reviewer can reconstruct the story later?

| **Goal** | **What you do** | **Why it matters** |
| --- | --- | --- |
| Reduce change | Limit who opens what; prefer working copies | Later hashes and reviews stay comparable |
| Reduce loss | Capture volatile or rotating sources early when authorized | Lost evidence cannot be reconstructed from hope |
| Document handling | Record who held the item, when, and why | Others can evaluate what was done |
| Keep baselines | Protect an original / master reference | Analysis mistakes do not erase the starting point |
| Separate roles | Intake vs analysis vs storage when possible | Reduces accidental overwrites and unclear possession |

**Easy example:** Instead of opening an unknown USB on a shared lab PC, label it, limit access, note the exterior, and examine a documented working copy under controlled conditions.

**Teaching walkthrough - decide first (use on every new item):**
1. Name the investigative question in one sentence.
2. List candidate relevant items for that question.
3. Rank write risk for each item (high / medium / low).
4. Choose the baseline form (device held sealed, export ZIP, or image - as authorized).
5. Name the first custodian and storage location.
6. Only then plan examination steps on working copies.

**🧩 Small Problem:** Is "I made a quick edit to fix formatting on the evidence file" acceptable?

**✅ Solved:** No. Do not alter the preserved original. Work on a documented working copy and leave the baseline untouched.

**🧩 Small Problem:** A teammate says, "We can preserve later - first let's see what files are on it." What is wrong?

**✅ Solved:** Previewing first can change the very state you meant to preserve. Identify, control, and document first; examine through an approved working-copy path.

---

### **1.2 Why Preservation Matters**

Preservation habits support fairness, repeatability, and accountability - even in campus training scenarios that never go to court.

Preservation is not packaging theater. It is the set of choices that keep later hashing, imaging, and reporting honest. When teams skip labels, open devices casually, or treat working edits as originals, every later technical step inherits that weakness.

| **Setting** | **What preservation protects** | **What failure looks like** |
| --- | --- | --- |
| Universities | USB sticks, exports, logs tied to academic integrity questions | Disputed "who opened it first" stories |
| Teaching labs | Comparable synthetic evidence packs across teams | Teams cannot reproduce each other's hashes |
| Incident drills | Avoidable change before analysis claims | Containment actions destroy needed records |
| Review boards | Ability to audit handling | Findings dismissed as unsupported process |

**Easy example:** A student says preservation is "only for court." The same habits still protect fairness in academic reviews and incident drills.

**Worked campus micro-case:**
A shared grades export is emailed to three TAs "so everyone can help." One TA sorts and saves; another filters and saves; the third hashes a different attachment version. Nobody can agree which file is the baseline. Preservation would have meant one master export, one hash, and working copies only.

**🧩 Small Problem:** A payment of effort goes into hashing, but nobody restricted who opened the USB. What is still weak?

**✅ Solved:** Hashing helps integrity checks, but uncontrolled access weakens reliability of any exclusive-control claim. Preservation includes access control and notes, not hashes alone.

**🧩 Small Problem:** Does a perfect hash after chaotic handling repair the chaos?

**✅ Solved:** No. A late hash fingerprints the late state. It does not prove earlier contents were unchanged during undocumented handling.

---

### **1.3 What Counts as Relevant Material**

**Definition:**
Relevant material is any data, device, or export that may answer an investigative question. Preservation starts by identifying what is in scope - not by collecting everything on campus.

💡 **In simple words:** *Preserve what the question needs. Ignoring scope either loses key items or floods the case with noise.*

| **Category** | **Campus-style examples (fictional)** | **Preservation note** |
| --- | --- | --- |
| Removable media | USB labeled "Exam Backup" | Avoid casual mounts; plan imaging/copy |
| Exported files | grades_source.csv, access_log.txt | Hash and store a master export |
| Device itself | Laptop left in a lab | Power state and network matter |
| Account exports | LMS submission ZIP | Keep original ZIP; analyze extracted copies |
| Scene photos / notes | Photo of USB on desk; sticky note on bag | Label time, author, and link to evidence ID |
| Derived products | Filtered CSV, screenshots | Label as derived; never replace the master |

**Scope mistakes to teach:**
* Collecting every USB in a room when only one labeled device is tied to the allegation.
* Preserving a laptop but forgetting the printed note taped to it that identifies ownership claims.
* Saving screenshots of a portal without recording which account and which URL produced them.
* Keeping analysis charts while deleting the only master export "to clean the folder."

**Relevance test (say it aloud):**
"If this item disappeared, could we still answer the investigation question with the remaining records?" If no, it is likely relevant and needs preservation attention.

**Easy example:** For "who edited the shared sheet after 5 PM?", relevant material may be the sheet export, edit history, and auth logs - not every file on the editor's drive.

**🧩 Small Problem:** A team preserves only a screenshot of a folder listing. What is missing?

**✅ Solved:** The listing is a derivative view. Prefer the original files or a verified export, plus notes on how the screenshot was made, by whom, and when.

**🧩 Small Problem:** Can metadata itself be relevant material?

**✅ Solved:** Yes. Timestamps, hashes, EXIF, and export headers can support or challenge claims. Preserve them with the same care as file contents.

## 🔹 **1.4 Write Risk and Controlled Access**

**Definition:**
Write risk is the chance that connecting, opening, or syncing an evidence item will change its contents or metadata. Controlled access means only authorized people handle the item, in documented ways, on suitable systems.

💡 **In simple words:** *Every casual open is a possible rewrite. Limit hands, limit machines, limit surprises.*

| **Action** | **Write risk** | **What may change** | **Safer teaching alternative** |
| --- | --- | --- | --- |
| Plug USB into personal laptop | High | Indexes, thumbs, autorun traces, timestamps | Approved lab workflow / write-blocker concepts (later weeks) |
| Open CSV in Excel and Save | High | Order, formatting, metadata | Copy first; open the copy |
| View PDF via browser downloads | Medium | Extra download copies; viewer caches | Work inside a labeled case folder |
| Sync folder to personal cloud | High | Renames, versioning, sharing side effects | Disable sync on evidence paths |
| "Quick antivirus repair" on media | High | Quarantine/delete/alter files | Do not "heal" evidence media in place |
| Read-only intent / forensic image (when taught) | Lower for master image | Still depends on correct process | Prefer imaged baseline for media when authorized |

**Access control checklist (teach as a habit):**
* Who may touch the physical item?
* Who may open the digital master?
* Where is the locker / encrypted store?
* Is the working environment free of auto-cleaners and sync clients?
* Are guests and classmates blocked from "just checking"?
* Is there a named person responsible right now?

**Easy example:** A print-lab USB sits overnight in an unlocked drawer. Even without deliberate edits, uncontrolled access creates a reliability problem for later exclusive-control claims.

**Instructor contrast table:**

| **Claim students want to make** | **Access reality that undermines it** |
| --- | --- |
| "Nobody else touched it" | Unlocked drawer / shared desk |
| "We only looked" | OS may still write |
| "It was with IT" | No named holder or time |

**🧩 Small Problem:** Is "we trusted everyone in the lab" a substitute for access notes?

**✅ Solved:** No. Trust is not a custody record. Write who held the item and where it was stored.

**🧩 Small Problem:** A read-only file attribute is set on Windows. Does that fully remove write risk for a USB?

**✅ Solved:** No. Attributes are not a complete forensic control. Hosts can still write other areas, and attributes can be cleared. Treat attribute tricks as incomplete protection.

## 🔹 **1.5 Contamination Risks**

**Definition:**
Contamination is unintended change or mixing that makes evidence harder to interpret or easier to challenge.

💡 **In simple words:** *Contamination is when your process adds footprints, mixes cases, or alters content without a clear record.*

**Common classroom examples:**
* Opening evidence on a machine that auto-updates files or generates thumbnails
* Sorting a CSV in place and saving over the export
* Mixing case folders so Case A files enter Case B
* Using screenshots without recording source and time
* Rebooting a live system before capturing volatile data when preservation was possible
* Running cleanup or antivirus "repair" on an evidence drive
* Copy-pasting between two open case folders until filenames collide
* Renaming the master to `final_clean.csv` after edits

| **Contamination type** | **What it looks like** | **First response** |
| --- | --- | --- |
| Content change | Hash differs; rows reordered | Restore from master; disclose the edit |
| Metadata change | Timestamps shifted after open | Note tool used; avoid treating new times as original |
| Cross-case mix | Wrong case ID on a file | Quarantine, re-label, reconstruct source |
| Process noise | Extra files from viewer caches | Keep analysis off the master path |
| Social contamination | Unlogged pass-around | Reconstruct holders; disclose residual gap |

**Easy example:** You open `grades_source.csv`, sort by name, and click Save. If that path was treated as the original, integrity is damaged.

**Recovery teaching sequence:**
1. Stop further edits.
2. Identify whether a true master still exists.
3. If yes, recreate a clean working copy and quarantine the contaminated file.
4. If no, disclose that the baseline itself changed and limit claims accordingly.
5. Update custody / notes with what happened - including the mistake.

**🧩 Small Problem:** You sorted an evidence CSV and clicked Save. What is the issue?

**✅ Solved:** The file changed. Restore from the preserved baseline, document the mistake, and continue only on a fresh working copy.

**🧩 Small Problem:** Two teams share one Desktop folder named `evidence`. Why is that risky?

**✅ Solved:** Filenames collide and ownership of edits becomes unclear - classic cross-contamination and custody confusion.

✅ **Opening Activity - Preserve or Contaminate?**

Classify each action:

1. Label USB as EV-014 and lock it in L2.
2. Open the only export in Excel and save a sorted version over it.
3. Copy `evidence.bin` to `evidence_working.bin` before parsing.
4. Email the only master to five classmates "for speed."
5. Photograph the exterior label and record the finder name.

**Discussion:** 1 and 3 and 5 preserve (or support preservation); 2 and 4 contaminate or create uncontrolled copies.

# **2. Integrity, Authenticity, and Working Copies**

---

## 🔹 **2.1 Integrity, Authenticity, and Reliability**

These three words are often mixed. Treat them as separate questions.

Teach them aloud as three checks in order: (1) Do the compared bytes match? (2) Is this the claimed item from the claimed source? (3) Is the process good enough for the specific claim being made?

| **Idea** | **Question it answers** | **Classroom check** |
| --- | --- | --- |
| **Integrity** | Were the bytes altered unexpectedly? | Matching SHA-256 before/after transfer |
| **Authenticity** | Is this the item claimed (right source, right export)? | Source path, intake label, export method |
| **Reliability** | Is the source and process trustworthy enough for this claim? | Known tool, complete fields, disclosed limits |

💡 **In simple words:** *Integrity asks "same bytes?" Authenticity asks "right item?" Reliability asks "can we trust this source for this question?"*

**How they interact (lecture map):**
* Matching hashes support integrity between compared copies. They do **not** by themselves prove authorship.
* A genuine USB with a broken custody story may still be authentic as an object, yet less reliable for strong overnight exclusive-control claims.
* A reliable process can still produce incomplete evidence; honesty about limits is part of reliability.
* Students often collapse all three into "the hash proves the case." Separate the questions every time.

**Easy example:** If two copies share the same SHA-256, their compared bytes match - that does not prove who authored the file.

| **Situation** | **Integrity** | **Authenticity** | **Reliability for "student X edited it"** |
| --- | --- | --- | --- |
| Hash match after controlled transfer | Supported | Still needs labels/source notes | Needs account/log linkage, not hash alone |
| Hash mismatch after casual open | Broken or unexplained | USB may still be authentic as an object | Weakened until change is explained |
| Correct file but wrong semester export | Bytes may be intact | Wrong item for the claim | Poor for this semester's question |
| Right export, incomplete custody overnight | May still match later hash | Item identity may be fine | Weak for exclusive-control claims |

**🧩 Small Problem:** Two files have different hashes. Does that prove one is forged?

**✅ Solved:** No. Different content, different versions, or transfer error can also explain a mismatch. Investigate the cause.

**🧩 Small Problem:** A hash matches, so a student claims the case is "proven." What is wrong?

**✅ Solved:** Integrity of compared bytes is not a full case proof. Authenticity of source and reliability of interpretation still need separate support.

**🧩 Small Problem:** Which fails first when classmates passed a USB around undocumented: today's matching hash, or last week's exclusive-control story?

**✅ Solved:** The exclusive-control story fails first. Today's hash can still be true while last week's control remains unproven.

## 🔹 **2.2 Original vs Working Copy**

**Definition:**
The original (or master) is the preserved evidence baseline. A working copy is a duplicate used for examination so the baseline stays unchanged.

💡 **In simple words:** *Lock the original. Analyze the copy. If the copy changes, the original still provides the reference.*

| **Item** | **Use** | **Typical controls** |
| --- | --- | --- |
| Original / master | Preserve, hash, store with restricted access | Read-only intent, sealed storage, limited holders |
| Working copy | Open, parse, filter, annotate findings | Derived from master; named clearly (`*_working`) |
| Analysis product | Notes, charts, filtered CSVs | Labeled as derived; not treated as the only master |
| Shareable excerpt | Redacted extract for discussion | Linked back to evidence ID; not a silent second master |

**Folder habit (classroom):**

```
EV-014/
  master/
    evidence.bin
    baseline_sha256.txt
  working/
    evidence_working.bin
  products/
    findings_notes.txt
```

**Naming rules that prevent mix-ups:**
* Never name the master `final.csv` after editing.
* Put `working` or `derived` in disposable filenames.
* Keep `baseline_sha256.txt` next to the master, not only in a chat message.
* If you must email something, email a clearly labeled working copy or product - and still keep the master where it belongs.

**Easy example:** Keep `grades_source.csv` untouched and edit only `grades_working.csv` during analysis practice.

**Common student shortcuts that break this model:**
* Hashing once, then editing the same path "because we already hashed it."
* Renaming the master after sorting.
* Emailing the only copy to three classmates without tracking who changed what.
* Saving Excel "compatibility" changes on top of the master export.

**🧩 Small Problem:** Should students edit the only copy of `grades.csv` "because we hashed it earlier"?

**✅ Solved:** No. Keep a preserved baseline. Edit only a working copy. An old hash does not protect a file you later overwrite.

**🧩 Small Problem:** Analysis crashes and corrupts the working copy. What do you do?

**✅ Solved:** Quarantine the damaged working file, recreate a fresh working copy from the master, and document the incident. Do not "fix" the master in place.

**🧩 Small Problem:** Your working copy is more "organized" than the master. Should you replace the master with it?

**✅ Solved:** No. Organization is an analysis product. Keep the master as the baseline; save the organized file under `products/` with a note that it is derived.

# **2.3 Seals, Hashes, and When to Re-Check**

---

### **Definition**

Seals and hashes are integrity controls. A seal shows whether a physical package was opened. A hash shows whether digital bytes changed.

💡 **In simple words:** *Seals watch the bag. Hashes watch the bits. You often need both stories.*

| **Control** | **Detects** | **Does not prove alone** |
| --- | --- | --- |
| Physical seal | Package opened / resealed poorly | Exact file contents inside |
| SHA-256 hash | Byte-level change in hashed object | Who changed it or why |
| Dual control / witness | Single-person silent access | Perfect honesty of witnesses |
| Read-only media claim | Intent to avoid writes | Actual host behavior |

**Practice points:**
* **Physical seal:** Tamper-evident sticker or bag with unique ID recorded in custody notes.
* **Hash:** SHA-256 (or similar) digest recorded at key handling points.
* **Re-check:** Recalculate after transfer or before analysis and compare to the baseline.
* **Scope the hash:** Hash the right object (image file, export ZIP, not a random preview).
* **Record algorithm + object path:** "SHA-256 of EV-014/master/evidence.bin" beats a bare hex string in chat.

**Easy example:** Seal S-19 is intact and `evidence.bin` still matches the intake SHA-256. That supports packaging unopened and bytes unchanged since intake - still not a verdict about who created the files.

---

### **How We Use Hashing in This Week's Logic**

1. Identify the exact object to preserve.
2. Calculate and record the baseline digest at intake (or at master creation).
3. Create a working copy from the master.
4. Before deep analysis, confirm the working baseline still matches when policy requires it.
5. If digests differ, investigate before blaming a person.

**Simple analogy:** A hash is a compact fingerprint of the compared bytes. The analogy does not mean the digest identifies the author or proves the story in the file is true.

---

### **When Digests Differ**

Possible reasons include wrong file path, incomplete copy, ordinary edits, line-ending changes, tool error, or deliberate alteration.

**Investigation order (classroom):**
1. Confirm path and filename.
2. Confirm algorithm (SHA-256 vs something else).
3. Confirm the full file was read (not a partial download).
4. Compare against master, not against another working draft.
5. Only then discuss handling history and possible alteration.

**🧩 Small Problem:** Seal intact but hash mismatch. What can you say?

**✅ Solved:** Physical packaging may be fine while the digital content differs from the recorded baseline. Investigate copy errors, wrong file, or alteration.

**🧩 Small Problem:** Hash matches but seal is broken. What can you say?

**✅ Solved:** Bytes of the hashed object match the baseline, yet physical access may have occurred. Document seal condition and evaluate other package contents.

---

### **When to Hash Again**

| **Event** | **Hash purpose** | **Typical object** |
| --- | --- | --- |
| Intake / receipt | Establish baseline | Device image, export set, or evidence file |
| After creating master image / preserved export | Confirm successful duplication | Master image or archive |
| Before analysis on a working baseline | Confirm copy still matches | Working copy vs master |
| After transfer (when policy requires) | Detect change in transit | Same hashed object |
| Before archive / return | Confirm package integrity | Final archived object |

💡 **In simple words:** *Hash when the evidence changes hands or when a new baseline is born - not when you merely read a working copy.*

**What re-hashing does not require:**
* Every mouse click during analysis on a disposable working file
* Every temporary in-memory sort that is never saved
* Every screenshot of a findings note

**Easy example:** Hash at intake, hash the master after acquisition, confirm the working copy before deep analysis. Do not re-hash after every spreadsheet filter.

**🧩 Small Problem:** Should you re-hash after every mouse click during analysis?

**✅ Solved:** No. Protect the master, work on copies, and re-hash when transfers or new baselines are created.

**🧩 Small Problem:** After transfer, hashes differ. First checks?

**✅ Solved:** Confirm same path/object, same algorithm, complete file - then investigate handling. Do not silently overwrite the baseline with the new digest.

**🧩 Small Problem:** Is hashing encryption?

**✅ Solved:** No. Hashing supports integrity comparison. Encryption protects confidentiality with keys. Neither identifies the editor by itself.

✅ **Predict Before the Lab:** Will appending one space to a working copy change its SHA-256 while leaving the master unchanged? For this exercise, the answer is **yes**.

# **3. Chain of Custody**

---

## 🔹 **3.1 What Chain of Custody Is**

**Definition:**

Chain of custody is the documented history of evidence possession, transfer, storage, and handling.

💡 **In simple words:** *If you cannot say who had it, when, and why, the custody story has a hole.*

It helps a reader answer **what item was handled, who handled it, when, and why**.

**Minimum continuous story:**
`Found/collected → each transfer (from, to, time, purpose) → storage → release for analysis → return`

**Why custody is not optional paperwork:**
Integrity checks answer "same bytes?" Custody answers "who could have changed them, and under what documented control?" Reviewers need both.

**Easy example:** A custody row should say who held `USB-DF01`, when it moved from A to B, and why - not only "IT has it somewhere."

**🧩 Small Problem:** Hash matches, but overnight holders are unknown. Is handling history complete?

**✅ Solved:** No. Hash comparison and custody history answer different questions. Document the missing period; do not fabricate a transfer record.

---

## 🔹 **3.2 Packaging, Labels, and Evidence IDs**

**Definition:**
Packaging and labels give physical and administrative identity to an item. An evidence ID is a unique tag used in every custody row, hash note, and findings file.

💡 **In simple words:** *If three USBs look alike, only the ID and label keep the story attached to the right object.*

| **Element** | **Good practice** | **Weak practice** |
| --- | --- | --- |
| Evidence ID | `EV-014-USB01` used everywhere | "the black USB" |
| Description | Capacity, label text, visible marks | "backup stick" |
| Seal / bag ID | Unique sticker recorded at sealing | Unmarked ziplock |
| Photo / sketch | Exterior label and ports (when used) | No visual reference |
| Storage location | Locker L2 shelf B | "somewhere in IT" |

**Label text should answer:**
1. What is this item?
2. Which case or exercise ID?
3. When was it bagged or registered?
4. Who sealed or registered it?

**Easy example:** Two fictional USBs both say "backup." Only `EV-014-USB01` vs `EV-015-USB01` keeps hash files and custody rows from being swapped.

**🧩 Small Problem:** Can you reuse evidence IDs across scenarios "to keep things simple"?

**✅ Solved:** No for real case discipline. Unique IDs prevent mix-ups. In class, still practice unique IDs per scenario.

**🧩 Small Problem:** The exterior label peels off. What do you document?

**✅ Solved:** Note the condition change, re-secure packaging, update the description, and keep the old ID. Do not invent a new identity without linking to the old one.

---

## 🔹 **3.3 Custody Fields and How to Write an Entry**

| **Field** | **Example value** | **What it proves to a reviewer** |
| --- | --- | --- |
| Evidence ID | EV-014-USB01 | Same object across notes |
| Description | 16 GB USB labeled "Exam Backup" (fictional) | Physical identity |
| Collected / found by | Examiner A / Student Finder | Intake source |
| Date/time collected | 2026-10-02 10:15 +05:00 | When control began |
| From / to | Desk drawer → Evidence locker L2 | Movement path |
| Purpose | Secure storage pending imaging | Why the move happened |
| Integrity check | SHA-256 recorded at intake | Byte baseline when controlled |
| Condition | Seal intact / ports clear | Physical state notes |

**Template (classroom):** `seq`, `evidence_id`, `from_holder`, `to_holder`, `action`, `timestamp` with offset, `purpose`, `sha256_at_event`, `notes`.

**Writing rules:**
* Prefer timestamps with timezone or offset (for example `+05:00`).
* Short true purpose beats empty purpose.
* Condition notes matter when seals, ports, or packaging change.
* Do not write conclusions about guilt in a custody row.

| **Weak** | **Stronger** |
| --- | --- |
| "Got the USB later" | "Received EV-014 from A at 11:00 +05:00 for secure storage in L2" |
| "Checked it" | "Verified seal S-19 intact; no ports opened; returned to locker L2" |
| "Maybe B had it" | "Gap: holder between 14:00 and 16:00 not documented; investigating" |
| "Transferred for analysis" with no names | "From B to C at 14:30 +05:00; purpose: create working copy; master returned to L2" |

**Easy example:** After IT receives a stick from faculty, the entry names both people, the time, the purpose ("intake for hashing"), and the storage shelf.

**🧩 Small Problem:** You forgot the exact minute of a handoff. What should you write?

**✅ Solved:** Write the best supported time window and mark uncertainty. Do not invent a precise minute.

**🧩 Small Problem:** Is it OK to silently backfill yesterday's missing rows from memory to make the CSV look complete?

**✅ Solved:** Only with clear labeling that entries are reconstructed, and only for what memory or other records truly support.

**🧩 Small Problem:** Must every person who glances at a screen get a custody row?

**✅ Solved:** Physical possession and control transfers need rows. Mere observation without handling may be a case note instead - but if someone takes the device away, that is a transfer.

# **3.4 Custody Gaps and Multi-Person Handoffs**

---

### **Definition**

A custody gap is a missing or broken link in the handling history, such as an unexplained holder change or undocumented time period.

💡 **In simple words:** *A gap means the story of who controlled the item is incomplete.*

| **Symptom** | **Why it matters** | **Repair approach** |
| --- | --- | --- |
| Holder B receives from A, but A never logged a transfer | Continuity unclear | Interview, reconstruct, disclose residual gap |
| Overnight storage with no location record | Access opportunity undocumented | Record known location facts; admit unknowns |
| Hash missing at intake | Later changes harder to detect | Hash now; treat pre-hash period as limited |
| Sequence jumps A→C while B was current | Broken continuity | Do not invent B→C; investigate real path |
| Temporary "watch this" hold unlogged | Silent access window | Add reconstructed note if supported; else disclose |

**Gap response protocol:**
1. Stop quietly inventing rows.
2. Write that a gap exists (who noticed, when).
3. Gather secondary sources if authorized (email, camera notes, witness statements).
4. Reconstruct only supported links.
5. Carry remaining uncertainty into findings language.

**Easy example:** Faculty hands a USB to IT, but the earlier student→faculty handoff was never written down. A late IT hash cannot erase that earlier undocumented period.

**Impact teaching table:**

| **Gap type** | **Claim most weakened** | **Claim that may still stand** |
| --- | --- | --- |
| Undocumented lunch hold | Exclusive control during lunch | Later hash match after return to locker |
| Missing intake hash | Pre-IT integrity | Description of current contents after hashing |
| Broken A→C jump | Continuous holder story | Whatever transfers are still documented |

**🧩 Small Problem:** What should you do when you discover a gap?

**✅ Solved:** Document the gap honestly, investigate what can be reconstructed, and do not invent missing transfers.

**🧩 Small Problem:** Does a one-hour undocumented locker period automatically destroy every finding?

**✅ Solved:** Not automatically. It weakens confidence for claims that depend on exclusive control during that hour. Disclose impact relative to the specific claim.

---

### **Handoff Discipline**

Handoff discipline means completing a custody entry when control moves - including temporary "can you hold this" moments.

| **Handoff type** | **Example** | **Required notes** |
| --- | --- | --- |
| Finder → first custodian | Student to lab officer | Found location, time, condition |
| Custodian → storage | Officer to locker manager | Location ID, seal if used |
| Storage → analyst | Locker to examiner | Purpose: analysis; hash check |
| Analyst → storage | Return after working copy made | Master returned; seal status |
| Temporary hold | Desk neighbor holds bag briefly | Still a transfer if control moved |

**Continuity check:** after each row, the next `from_holder` should match the previous `to_holder`. If not, you have a continuity break to investigate - the same idea the lab gap detector will practice.

**Easy example:** A asks B to watch EV-014 during lunch and never logs it. Reviewers later treat A as continuous holder while B actually had access - a silent gap.

**🧩 Small Problem:** A emails B "USB is in my drawer if you need it" and B takes it without a meeting. What is missing?

**✅ Solved:** A documented transfer with time, both names, purpose, and condition. Email intent is not a completed custody row.

**🧩 Small Problem:** Can two people be "joint holders" without a row?

**✅ Solved:** Joint control should still be written (both names, shared location, rules). Unwritten joint access is a gap waiting to happen.

## 🔹 **4.1 Handling Scenario - USB in the Print Lab**

**Fictional case:** A USB labeled "Exam Backup" is found in lab W08 after midterm printing. Campus security asks the forensics teaching team to preserve it for a training exercise modeled on a real workflow.

**Sound handling outline:**
1. Photograph or note label text and ports area (training photos optional).
2. Assign evidence ID and write a clear description.
3. Record who found it, who received it, date/time with offset, and location.
4. Limit further openings; do not browse on random PCs.
5. Create a forensic image or bit-for-bit working copy when authorized (detailed imaging is a later week topic).
6. Hash the preserved image/file set and store the master securely.
7. Analyze only working copies.
8. Keep custody rows updated for every transfer.
9. Close with return/archive notes when the exercise ends.

| **Choice** | **Prefer** | **Avoid** |
| --- | --- | --- |
| Identify | Unique ID + label notes | "that stick from the lab" |
| Access | Authorized holder only | Pass-around "to see whose it is" |
| Examine | Documented working copy / image | Personal laptop preview |
| Record | Intake hash + custody | "hash later if needed" only |
| Communicate | Facts and next steps | Speculating about guilt in intake notes |

**Timed decision drill (discussion, not a stopwatch exercise):**
* Security wants a verbal update before paperwork - protect physical control and start written intake first.
* A classmate recognizes the label style - still do not open files to "confirm ownership."
* IT offers to "just plug it in the helpdesk PC" - decline uncontrolled mounts.

**Easy example:** Student A finds the stick and wants to open two PDFs "to see whose it is." That curiosity creates write risk and an undocumented viewing event.

**🧩 Small Problem:** Is plugging the USB into a personal laptop "just to see files" good preservation?

**✅ Solved:** No. That risks writes and contamination. Use a controlled, documented process.

**🧩 Small Problem:** Security wants a verbal update before any paperwork. What do you protect first?

**✅ Solved:** Keep physical control and start written intake (ID, finder, time, location). Verbal updates can follow; undocumented possession should not.

## 🔹 **4.2 Preservation Mistakes to Avoid**

| **Mistake** | **Why it hurts** | **Better habit** |
| --- | --- | --- |
| Editing the only copy | No clean baseline remains | Master + working copy |
| Skipping intake notes | Later custody reconstruction fails | ID, time, finder, location at first control |
| Mixing case folders | Cross-contamination risk | Separate trees per evidence ID |
| Hiding a gap | Reviewers lose trust in the whole record | Disclose and investigate |
| Hashing the wrong file | False confidence | Verify path and filename before hashing |
| Previewing on random hosts | Uncontrolled writes | Approved workflow only |
| Over-claiming from a late hash | Late hash ≠ early integrity | State what the hash actually covers |
| Vague labels | Swap risk | Unique IDs and descriptions |
| Silent backfill of custody rows | Fake completeness | Mark reconstructed entries openly |

**Easy example:** A team skips intake notes because "everyone saw who found the USB." Later, memories disagree. A one-line intake note would have prevented the dispute.

**Mistake autopsy (teach with one fictional failure):**
Team opens USB on three laptops, finally hashes a fourth copy, then writes "chain of custody complete." A reviewer asks for holders between discovery and hashing. Silence. The technical hash work cannot repair the missing story.

## 🔹 **4.3 Reliability vs Perfection**

💡 **In simple words:** *Preservation aims for disciplined handling, not magical certainty. Document limits honestly.*

* A complete custody register supports reliability.
* A disclosed gap is better than a fabricated link.
* A hash mismatch is information; investigate it.
* Classroom synthetic files train the habits without real privacy risk.
* Strong process still allows incomplete evidence - say what you can and cannot support.

| **Claim strength** | **Needs** | **Weakened by** |
| --- | --- | --- |
| "Bytes match intake baseline" | Matching hashes + correct object | Hashing wrong file; silent edits |
| "Item stayed under listed holders" | Continuous custody rows | Overnight undocumented access |
| "Student X authored content" | Authorship sources beyond custody | Custody alone |
| "No one could have changed it overnight" | Controlled storage + access evidence | Unlocked drawer; missing location |

**Easy example:** Your register shows a two-hour gap. Hashes still match after the gap. You may support byte integrity after re-establishment of control, while remaining cautious about exclusive control during the gap.

**🧩 Small Problem:** Is a disclosed custody gap automatically fatal to every finding?

**✅ Solved:** Not automatically. It weakens confidence and must be reported; impact depends on the gap and the claim.

**🧩 Small Problem:** Does perfection mean zero uncertainty in notes?

**✅ Solved:** No. Perfection here means honest, complete documentation of what is known and unknown - not fake certainty.

## 🔹 **4.4 Documentation Language that Survives Review**

**Definition:**
Review-safe language states observations, methods, and limits without leaping to guilt, motive, or unsupported certainty.

💡 **In simple words:** *Write like someone else will challenge every adjective.*

| **Avoid** | **Prefer** |
| --- | --- |
| "The student clearly tampered with the USB" | "Working copy hash differed from intake baseline after undocumented open; alteration cause not yet established" |
| "Chain of custody was fine" | "Custody rows continuous from intake through locker L2; no transfers logged overnight" |
| "Hash proves innocence" | "SHA-256 of master matched baseline at time T" |
| "Nobody could have accessed it" | "Access control relied on locker L2; badge logs not reviewed in this exercise" |
| "IT messed with evidence" | "IT held EV-014 from 14:30; no pre-receipt hash recorded" |

**Three labels to practice on every contested sentence:**
* **Observation:** what was seen or measured
* **Inference:** what you conclude from observations
* **Unknown:** what the records do not show

**Rewrite workshop (do in pairs):**

| **Draft sentence** | **Better version** |
| --- | --- |
| "Obviously contaminated" | "File order differs from intake export; save occurred on lab PC-12 at 11:40 +05:00 per note" |
| "Proven untouched" | "Seal S-19 intact at 15:00; SHA-256 matched intake baseline" |
| "Owner is Student A" | "Exterior label reads 'A. Khan backup'; ownership not established by label alone" |

**Easy example:** Observation: "Seal S-19 broken at 14:05." Inference: "Package may have been opened." Unknown: "Who opened it and whether files changed before re-hash."

**🧩 Small Problem:** Rewrite: "Obviously the IT guy messed with the evidence."

**✅ Solved:** Example: "IT assistant held EV-014 from 14:30; no hash was recorded before that receipt. Pre-receipt integrity cannot be confirmed from available notes."

**🧩 Small Problem:** Should findings files include opinions about punishment?

**✅ Solved:** No. Preservation and custody notes support technical and procedural facts. Outcomes belong to authorized decision-makers.

## 🔹 **4.5 Storage, Retention, and Return**

**Definition:**
Storage is how and where the master lives while waiting for analysis. Retention is how long it stays under control. Return or release is how custody ends when the exercise or case closes.

💡 **In simple words:** *Preservation does not end at intake. Where it sleeps, how long it stays, and how it leaves all need notes.*

| **Phase** | **What to record** | **Failure mode** |
| --- | --- | --- |
| Storage | Location ID, access list, seal status | "In IT somewhere" |
| Retention | Why still held; review date if policy uses one | Forgotten items with no owner |
| Interim checks | Spot seal/hash checks when required | Silent long custody with no verification |
| Return / archive | Who received it back; final hash; open gaps | Item vanishes from the register at the end |

**Campus training pattern:**
1. Master stays in a labeled folder or locker concept (`EV-014/master`).
2. Working copies live separately and can be deleted after the lab if policy allows.
3. Findings products are not substitutes for the master.
4. When the scenario ends, write a closing row: archived, returned, or destroyed (synthetic only).

**Easy example:** After analysis, examiner C returns the sealed bag to locker L2 and logs `action=return`, seal intact, final hash match.

**🧩 Small Problem:** Analysis is done, but the master USB sits on an analyst's desk for days. What is wrong?

**✅ Solved:** Unlogged desk storage is uncontrolled access. Return it to recorded storage or log the desk as the temporary location with holder name.

**🧩 Small Problem:** Can you delete the master after hashing "to save space"?

**✅ Solved:** Not if later review may need the baseline. Keep the master per retention rules; delete disposable working copies instead.

## ** Setup - Synthetic Evidence Item**

Create a temporary evidence file, a baseline hash, and starter custody rows for a fictional USB export.

````python
# Objective: Create a synthetic evidence file and starter custody rows for Week 3.
# Output: Temporary evidence.bin, baseline SHA-256, and custody_seed.csv.
# Interpretation: Setup gives a clean baseline before any handling demo.
# Forensic Meaning: Real cases begin with identified evidence and an intake hash whenever feasible.

from pathlib import Path
# Import tempfile for this step.
import tempfile
# Import hashlib for this step.
import hashlib
# Import csv for this step.
import csv
# Import json for this step.
import json
# Import datetime for this step.
from datetime import datetime, timezone, timedelta

# Create a fresh temporary folder so personal files are not touched.
lab_dir = Path(tempfile.mkdtemp(prefix="digital_forensics_week3_"))
# Set evidence_path for the next steps.
evidence_path = lab_dir / "evidence.bin"

# Fixed bytes so every student starts from the same baseline content.
evidence_path.write_bytes(b"GREENFIELD-USB-EXPORT-v1\nexam_midterm.pdf\n")

# Define helper function sha256_file.
def sha256_file(path: Path) -> str:
    # Full-file digest used as an integrity fingerprint.
    return hashlib.sha256(path.read_bytes()).hexdigest()

# Compute or compare a SHA-256 integrity digest.
baseline = sha256_file(evidence_path)
# Write synthetic training bytes/text into the file.
(lab_dir / "baseline_sha256.txt").write_text(baseline + "\n", encoding="utf-8")

# Seed custody events with continuous holders A -> B -> C.
PKT = timezone(timedelta(hours=5))
# Set seed_rows for the next steps.
seed_rows = [
    # Perform this step of the practical.
    {"seq": "1", "evidence_id": "EV-014", "from_holder": "FINDER", "to_holder": "A", "action": "intake", "timestamp": "2026-10-02T10:15:00+05:00"},
    # Perform this step of the practical.
    {"seq": "2", "evidence_id": "EV-014", "from_holder": "A", "to_holder": "B", "action": "transfer", "timestamp": "2026-10-02T11:00:00+05:00"},
    # Perform this step of the practical.
    {"seq": "3", "evidence_id": "EV-014", "from_holder": "B", "to_holder": "C", "action": "transfer", "timestamp": "2026-10-02T14:30:00+05:00"},
]

# Set seed_path for the next steps.
seed_path = lab_dir / "custody_seed.csv"
# Open a file/context safely for this step.
with seed_path.open("w", newline="", encoding="utf-8") as f:
    # Set writer for the next steps.
    writer = csv.DictWriter(f, fieldnames=list(seed_rows[0].keys()))
    # Perform this step of the practical.
    writer.writeheader()
    # Perform this step of the practical.
    writer.writerows(seed_rows)

# Show the result for discussion.
print("Training folder:", lab_dir)
# Show the result for discussion.
print("Baseline SHA-256:", baseline[:16] + "...")
# Show the result for discussion.
print("Seed transfers:", len(seed_rows))
````

````python
# Objective: Build a custody register with purpose and hash columns.
# Output: custody_register.csv including holder sequence and intake hash.
# Interpretation: A complete register makes later gap checks possible.
# Forensic Meaning: Custody documentation supports authenticity and handling transparency.

register_rows = []
# Open a file/context safely for this step.
with seed_path.open(encoding="utf-8") as f:
    # Loop through each item needed for the check.
    for row in csv.DictReader(f):
        # Attach purpose and the intake baseline hash to each logged transfer.
        register_rows.append(
            # Perform this step of the practical.
            {
                # Perform this step of the practical.
                **row,
                # Perform this step of the practical.
                "purpose": "secure_storage" if row["action"] == "intake" else "controlled_transfer",
                # Compute or compare a SHA-256 integrity digest.
                "sha256_at_event": baseline,
            }
        )

# Set register_path for the next steps.
register_path = lab_dir / "custody_register.csv"
# Set fields for the next steps.
fields = list(register_rows[0].keys())
# Open a file/context safely for this step.
with register_path.open("w", newline="", encoding="utf-8") as f:
    # Set writer for the next steps.
    writer = csv.DictWriter(f, fieldnames=fields)
    # Perform this step of the practical.
    writer.writeheader()
    # Perform this step of the practical.
    writer.writerows(register_rows)

# Set holders for the next steps.
holders = " -> ".join(r["to_holder"] for r in register_rows)
# Show the result for discussion.
print("Holder sequence:", "FINDER -> " + holders)
# Show the result for discussion.
print("Saved:", register_path.name)
````

````python
# Objective: Detect holder-continuity gaps in custody sequences.
# Output: No issues for the original register; issue listed for a broken sequence.
# Interpretation: A gap means the documented handoff story is incomplete.
# Forensic Meaning: Undocumented possession periods weaken confidence in handling history.

def find_custody_gaps(rows):
    # Set issues for the next steps.
    issues = []
    # Current holder starts as the first to_holder after intake from FINDER.
    current = None
    # Loop through each item needed for the check.
    for row in rows:
        # Branch on the condition relevant to this check.
        if current is None:
            # Set current for the next steps.
            current = row["to_holder"]
            # Perform this step of the practical.
            continue
        # Each transfer must leave from the person who currently holds the item.
        if row["from_holder"] != current:
            # Store this item in the working list.
            issues.append(
                # Perform this step of the practical.
                f"Gap before seq {row['seq']}: expected from_holder={current}, found {row['from_holder']}"
            )
        # Set current for the next steps.
        current = row["to_holder"]
    # Return the computed result to the caller.
    return issues

# Set good_issues for the next steps.
good_issues = find_custody_gaps(register_rows)
# Show the result for discussion.
print("Good register issues:", good_issues or "none")

# Broken story: B transfers to C even though current holder is still A.
broken = [
    # Perform this step of the practical.
    register_rows[0],
    # Perform this step of the practical.
    {**register_rows[1], "from_holder": "B", "to_holder": "C", "seq": "2"},
]
# Set broken_issues for the next steps.
broken_issues = find_custody_gaps(broken)
# Show the result for discussion.
print("Broken register issues:")
# Loop through each item needed for the check.
for issue in broken_issues:
    # Show the result for discussion.
    print("-", issue)

# Write synthetic training bytes/text into the file.
(lab_dir / "gap_report.json").write_text(
    # Convert between Python objects and JSON text.
    json.dumps({"good": good_issues, "broken": broken_issues}, indent=2),
    # Set encoding for the next steps.
    encoding="utf-8",
)
````

### **✅ Interpretation**

* Continuous from/to holders produce no gap issues.
* Jumping holders without a matching release creates a detectable gap.
* In real work, investigate and disclose gaps; do not invent missing rows.
