## 🔹 **1. Why Planning Comes First**

**Definition:**

Investigation planning decides what questions to answer, what is in scope, what evidence sources matter, and what authorization is needed before collection begins.

💡 **In simple words:** *Do not collect first and invent the plan later. Decide the questions, limits, and sources, then gather only what those questions need.*

**Easy example:** A complaint says "the exam leaked." A better first question is: "Which accounts downloaded exam_v3.pdf between 09:00 and 10:00 PKT?"

| **Without a plan** | **With a plan** |
| --- | --- |
| Random downloads and screenshots | Targeted, authorized collection |
| Unclear goals | Clear investigative questions |
| Extra privacy risk | Scoped access to relevant sources |

**🧩 Small Problem:** A manager says, "Check everything on the student's laptop." Should you start imaging immediately?

**✅ Solved:** No. First clarify the allegation, the questions, the authorized scope, and which sources can answer those questions.

## 🔹 **2. Investigation Triggers**

**Definition:**

A trigger is the event or report that starts an inquiry, such as a complaint, alert, audit finding, or policy concern.

💡 **In simple words:** *Something happened or was reported. That report starts the process; it is not yet a proven fact.*

**Easy example:** In a simple campus case, use one record for "Investigation Triggers" and say what it supports versus what still needs another source.

| **Trigger type** | **Example (fictional)** |
| --- | --- |
| Complaint | A student reports a missing graded file. |
| System alert | Unusual download volume on a shared drive. |
| Audit finding | Access log shows after-hours exam-folder access. |
| Policy concern | A USB device was used in a restricted lab. |

**🧩 Small Problem:** Is a trigger the same as proof that misconduct occurred?

**✅ Solved:** No. A trigger opens the inquiry. Facts come from authorized, relevant records.

✅ **Activity:** Write one trigger sentence for a fictional "shared folder modified overnight" report. Keep it free of blame language.

## 🔹 **3. Allegation vs Facts**

**Definition:**

An allegation is a claim that something improper happened. A fact is an observation supported by a reliable record or examination.

💡 **In simple words:** *"Someone stole my file" is an allegation. "Audit log shows account A downloaded F204 at 09:05" is closer to a recordable fact.*

**Easy example:** Trigger: a TA reports a strange download alert. Fact so far: an alert exists. Not yet a fact: "TA B leaked the file."

| **Statement** | **Type** | **Why** |
| --- | --- | --- |
| "Staff17 leaked the paper." | Allegation | Names a person and motive without proof. |
| "Account staff17 downloaded exam.pdf at 09:05." | Fact candidate | Tied to a logged action. |
| "Therefore staff17 intended to cheat." | Inference | Needs more support; intent is not automatic. |

**🧩 Small Problem:** A screenshot shows a chat message saying "I have the paper." Is the allegation proven?

**✅ Solved:** Not yet. Verify authenticity, account ownership, time, and whether the file was actually transferred.

## 🔹 **4. Investigative Questions**

**Definition:**

Investigative questions are focused questions the examination must answer. Good questions are answerable from digital sources and avoid assuming guilt.

**Easy example:** In a simple campus case, use one record for "Investigative Questions" and say what it supports versus what still needs another source.

**Strong question patterns:**

* What accounts accessed folder X between T1 and T2?
* Which devices connected to workstation W08 that morning?
* Was file F204 downloaded, copied, or only opened?
* What remains unknown after examining available sources?

**🧩 Small Problem:** Rewrite: "Prove that Ali cheated." into a better investigative question.

**✅ Solved:** Example: "Which accounts accessed the exam folder between 08:00 and 10:00, and what actions were logged for file exam.pdf?"

✅ **Activity:** Write three investigative questions for a fictional missing coursework file. Mark which source might answer each one.

## 🔹 **5. Scope: In and Out**

**Definition:**

Scope states what systems, accounts, time windows, and data types are included, and what is excluded.

💡 **In simple words:** *Scope is the fence around the investigation. Inside the fence you examine. Outside the fence you do not go without new authorization.*

**Easy example:** A complaint says "the exam leaked." A better first question is: "Which accounts downloaded exam_v3.pdf between 09:00 and 10:00 PKT?"

| **In scope (example)** | **Out of scope (example)** |
| --- | --- |
| Shared drive folder Exam_Fall2026 | Personal social media accounts |
| Accounts with folder access rights | Unrelated department servers |
| 2026-10-01 00:00 to 2026-10-02 23:59 | Activity from last semester |
| Cloud audit export for that folder | Full phone imaging of every student |

**🧩 Small Problem:** Why list exclusions explicitly?

**✅ Solved:** Exclusions prevent mission creep, protect privacy, and keep the work tied to authorization.

## 🔹 **6. Stakeholders**

**Definition:**

Stakeholders are people or roles who authorize the work, provide access, receive findings, or are affected by the inquiry.

| **Role** | **Interest** |
| --- | --- |
| Requester / complainant | Wants the issue addressed |
| Authorizing official | Grants lawful access limits |
| IT / system owner | Provides logs and technical access |
| Investigator / examiner | Plans and documents the work |
| Privacy / legal advisor (as available) | Checks authorization and sensitivity |

💡 **In simple words:** *Know who can say yes, who can give you logs, and who receives the report. Do not treat every curious observer as a decision maker.*

**Easy example:** In a simple campus case, use one record for "Stakeholders" and say what it supports versus what still needs another source.

**🧩 Small Problem:** A classmate asks for a live demo of another student's mailbox "for learning." Is that okay?

**✅ Solved:** No. Classroom exercises use synthetic data. Real accounts need proper authorization and a real case.

## 🔹 **7. Evidence Source Planning**

**Definition:**

Evidence source planning maps each investigative question to candidate sources, expected record types, and practical access paths.

**Easy example:** A complaint says "the exam leaked." A better first question is: "Which accounts downloaded exam_v3.pdf between 09:00 and 10:00 PKT?"

| **Question** | **Candidate source** | **Record type** |
| --- | --- | --- |
| Who accessed the folder? | Cloud / file-server audit log | Account, action, time, path |
| Was a USB used? | Workstation connection log | Device ID, connect time |
| Was the file changed? | Version history / checksum baseline | Version ID, hash, editor |
| Was content shared externally? | Sharing / export audit | Recipient, share link, time |

## 🔹 **8. Legal and Authorization Awareness (Classroom Level)**

**Definition:**

Authorization is documented permission to access specific systems or data for a defined purpose and scope.

💡 **In simple words:** *In class we use fictional cases. In real work you do not collect first and ask permission later.*

**Easy example:** In a simple campus case, use one record for "Legal and Authorization Awareness (Classroom Level)" and say what it supports versus what still needs another source.

**Classroom rules of thumb:**

* Know who authorized the examination and what they authorized.
* Stay inside the written scope and time window.
* Prefer institutional exports over informal screenshots when possible.
* Record what you received, when, and from whom.
* Stop and escalate if a request exceeds the authorization.

**🧩 Small Problem:** IT offers "full mailbox dump of every student in the course" for a one-file complaint. What is the planning issue?

**✅ Solved:** The request is broader than the complaint. Seek a narrower, authorized source set tied to the investigative questions.

## 🔹 **9. Prioritization**

**Definition:**

Prioritization orders work by volatility, relevance, authorization readiness, and investigative value.

**Easy example:** In a simple campus case, use one record for "Prioritization" and say what it supports versus what still needs another source.

| **Priority cue** | **Why it matters** |
| --- | --- |
| Volatile sources | Memory and live sessions change or disappear quickly. |
| High relevance | Sources that can answer the main questions first. |
| Access ready | Authorized exports available now beat blocked ideal sources. |
| Integrity risk | Sources that may be overwritten soon need early preservation. |

**🧩 Small Problem:** Should you delay a cloud audit export because a full disk image "would be nicer"?

**✅ Solved:** Usually no. Take the authorized, relevant export while it is available, then plan heavier acquisition if justified and authorized.

## 🔹 **12. Common Planning Mistakes**

**Easy example:** A complaint says "the exam leaked." A better first question is: "Which accounts downloaded exam_v3.pdf between 09:00 and 10:00 PKT?"

| **Mistake** | **Better habit** |
| --- | --- |
| Collect everything first | Map questions to sources first |
| Treat allegation as fact | Separate claim language from records |
| Skip exclusions | Write out-of-scope items explicitly |
| Ignore retention timers | Prioritize time-limited exports |
| Expand quietly | Stop and seek new authorization |

**🧩 Small Problem:** A teammate adds a student's personal email "because it might help." What is wrong?

**✅ Solved:** It expands scope without authorization and is not tied to an approved investigative question.

## 🔹 **13. Plan Quality Checklist**

Use this checklist before collection:

1. Trigger recorded without blame language?
2. At least three investigative questions written?
3. Scope in and scope out both listed?
4. Each question mapped to a candidate source?
5. Authorizing stakeholder identified?
6. First priority source justified (volatility or retention)?

💡 **In simple words:** *If any checkbox fails, fix the plan before you touch evidence sources.*

```python
# Objective: Create a question-to-source matrix for the investigation plan.
# Output: source_matrix.csv linking questions to candidate sources and record types.
# Interpretation: If a question has no source, the plan is incomplete.
# Forensic Meaning: Source planning shows what can be examined and what remains unknown.

# Pair each question with a realistic synthetic source.
matrix_rows = [
    # Perform this step of the practical.
    {
        # Perform this step of the practical.
        "question_id": "Q1",
        # Perform this step of the practical.
        "question": plan["investigative_questions"][0],
        # Perform this step of the practical.
        "candidate_source": "cloud_share_Exam_Fall2026",
        # Perform this step of the practical.
        "record_type": "audit_log",
        # Perform this step of the practical.
        "authorized": "yes",
    },
    # Perform this step of the practical.
    {
        # Perform this step of the practical.
        "question_id": "Q2",
        # Perform this step of the practical.
        "question": plan["investigative_questions"][1],
        # Perform this step of the practical.
        "candidate_source": "cloud_share_Exam_Fall2026",
        # Perform this step of the practical.
        "record_type": "file_access_events",
        # Perform this step of the practical.
        "authorized": "yes",
    },
    # Perform this step of the practical.
    {
        # Perform this step of the practical.
        "question_id": "Q3",
        # Perform this step of the practical.
        "question": plan["investigative_questions"][2],
        # Perform this step of the practical.
        "candidate_source": "workstation_W08_connection_log",
        # Perform this step of the practical.
        "record_type": "usb_and_session_events",
        # Perform this step of the practical.
        "authorized": "yes",
    },
]

# Set matrix_path for the next steps.
matrix_path = lab_dir / "source_matrix.csv"
# Open a file/context safely for this step.
with matrix_path.open("w", newline="", encoding="utf-8") as f:
    # Set fields for the next steps.
    fields = ["question_id", "question", "candidate_source", "record_type", "authorized"]
    # Set writer for the next steps.
    writer = csv.DictWriter(f, fieldnames=fields)
    # Perform this step of the practical.
    writer.writeheader()
    # Perform this step of the practical.
    writer.writerows(matrix_rows)

# Display a compact view for discussion.
for row in matrix_rows:
    # Show the result for discussion.
    print(f"{row['question_id']}: {row['candidate_source']} ({row['record_type']})")
# Show the result for discussion.
print("Saved:", matrix_path.name)
```

## **3️⃣ Practical - Scope Checker**

### **📂 Tasks**

1. Compare proposed collections to authorized systems.
2. Flag any out-of-scope item.
3. Print accept / reject decisions.


```python
# Objective: Check proposed collection items against authorized systems in the brief.
# Output: Accept/reject decisions for each proposed item, including one rejection.
# Interpretation: A rejected item means stop and seek new authorization, not silent expansion.
# Forensic Meaning: Scope control protects privacy and keeps the examination defensible.

# Authorized systems come from the case brief, not from wish lists.
authorized = set(brief["authorized_systems"])

# Set decisions for the next steps.
decisions = []
# Open a file/context safely for this step.
with prop_path.open(encoding="utf-8") as f:
    # Loop through each item needed for the check.
    for row in csv.DictReader(f):
        # Set ok for the next steps.
        ok = row["source"] in authorized
        # Store this item in the working list.
        decisions.append(
            # Perform this step of the practical.
            {
                # Perform this step of the practical.
                "item_id": row["item_id"],
                # Perform this step of the practical.
                "source": row["source"],
                # Perform this step of the practical.
                "decision": "ACCEPT" if ok else "REJECT_OUT_OF_SCOPE",
            }
        )

# Persist decisions for the findings cell.
decisions_path = lab_dir / "scope_decisions.json"
# Write synthetic training bytes/text into the file.
decisions_path.write_text(json.dumps(decisions, indent=2), encoding="utf-8")

# Loop through each item needed for the check.
for d in decisions:
    # Show the result for discussion.
    print(f"{d['item_id']}: {d['decision']} <- {d['source']}")
```

### **✅ Interpretation**

* ACCEPT means the item is inside the authorized system list in this exercise.
* REJECT means stop; do not collect personal social media without new authorization.
* Scope checking is a control, not a finding about misconduct.

Summarize the plan quality and the scope result in cautious language.
